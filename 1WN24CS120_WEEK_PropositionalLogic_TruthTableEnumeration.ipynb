```python
from itertools import product

print("P Q R | P<->Q | Q or R | not P -> R | KB")

models = []

for P, Q, R in product([True, False], repeat=3):
    s1 = (P == Q)
    s2 = Q or R
    s3 = (not P) <= R

    KB = s1 and s2 and s3

    print(
        int(P), int(Q), int(R), "|",
        int(s1), "     |",
        int(s2), "      |",
        int(s3), "          |",
        int(KB)
    )

    if KB:
        models.append((P, Q, R))

print("\nModels of KB:", models)

print("KB entails Q:", all(Q for P, Q, R in models))
print("KB entails P -> Q:", all((not P or Q) for P, Q, R in models))
print("KB entails R -> Q:", all((not R or Q) for P, Q, R in models))
print("hrishikesh shet-1WN24CS120")
```
