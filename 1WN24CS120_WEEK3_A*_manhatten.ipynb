```python
import heapq

# Initial state
initial = (1, 2, 3,
           4, 0, 6,
           7, 5, 8)

# Take goal state from user
print("name: Hrishikesh shet[1WN24CS120]")
print("Enter goal state (9 numbers, use 0 for blank):")
goal = tuple(map(int, input().split()))

# Manhattan Distance
def h(state):
    distance = 0

    for i in range(9):
        if state[i] != 0:
            # Current position
            r1, c1 = divmod(i, 3)

            # Goal position
            j = goal.index(state[i])
            r2, c2 = divmod(j, 3)

            distance += abs(r1 - r2) + abs(c1 - c2)

    return distance


# Generate possible moves
def neighbors(state):
    pos = state.index(0)
    row, col = divmod(pos, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_pos = r * 3 + c
            new_state = list(state)

            new_state[pos], new_state[new_pos] = \
                new_state[new_pos], new_state[pos]

            yield tuple(new_state)


# A* Algorithm
def astar():
    pq = []

    # f(n) = g(n) + h(n)
    heapq.heappush(pq, (h(initial), 0, initial))

    parent = {initial: None}
    cost = {initial: 0}

    while pq:
        f, g, state = heapq.heappop(pq)

        if state == goal:
            path = []

            while state:
                path.append(state)
                state = parent[state]

            return path[::-1]

        for next_state in neighbors(state):
            new_cost = g + 1

            if next_state not in cost or new_cost < cost[next_state]:
                cost[next_state] = new_cost

                f = new_cost + h(next_state)

                heapq.heappush(pq, (f, new_cost, next_state))
                parent[next_state] = state

    return None


# Solve
path = astar()

if path:
    print("\nSolution:")

    for i, state in enumerate(path):
        print("Step", i)

        for j in range(0, 9, 3):
            print(state[j:j+3])

        print()

else:
    print("No solution found.")
```
