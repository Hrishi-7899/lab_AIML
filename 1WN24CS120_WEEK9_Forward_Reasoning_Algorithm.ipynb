```python

# Forward Chaining

def is_var(x):
    return isinstance(x, str) and x.startswith("?")

def subst(x, theta):
    while is_var(x) and x in theta:
        x = theta[x]
    return x

def unify(a, b, theta):
    if is_var(a):
        theta = theta.copy()
        theta[a] = b
        return theta
    if is_var(b):
        theta = theta.copy()
        theta[b] = a
        return theta
    if a == b:
        return theta
    return None

def match(p, f, theta):
    if p[0] != f[0]:
        return None

    for x, y in zip(p[1:], f[1:]):
        theta = unify(subst(x, theta), subst(y, theta), theta)
        if theta is None:
            return None

    return theta

def forward_chaining(facts, rules, goal):

    while True:
        new = []

        for conditions, result in rules:
            substitutions = [{}]

            for condition in conditions:
                temp = []

                for theta in substitutions:
                    for fact in facts:
                        t = match(
                            tuple(subst(x, theta) for x in condition),
                            fact, theta
                        )
                        if t is not None:
                            temp.append(t)

                substitutions = temp

            for theta in substitutions:
                fact = tuple(subst(x, theta) for x in result)

                if fact not in facts and fact not in new:
                    new.append(fact)

        if not new:
            break

        facts.extend(new)
        print("New Facts:", new)

        if goal in facts:
            return True

    return goal in facts


# Initial Facts
facts = [
    ("Parent", "Alice", "Bob"),
    ("Parent", "Bob", "Charlie")
]

# Rules
rules = [
    ([("Parent", "?x", "?y")],
     ("Ancestor", "?x", "?y")),

    ([("Parent", "?x", "?y"),
      ("Ancestor", "?y", "?z")],
     ("Ancestor", "?x", "?z"))
]

# Goal
goal = ("Ancestor", "Alice", "Charlie")

print("Initial Facts:", facts)
print("Goal:", goal)

if forward_chaining(facts, rules, goal):
    print("Goal is TRUE")
else:
    print("Goal is FALSE")
```
