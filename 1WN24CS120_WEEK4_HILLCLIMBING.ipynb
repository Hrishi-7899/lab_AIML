```python

import random

def cost(b):
    return sum(b[i] == b[j] or abs(b[i]-b[j]) == j-i
               for i in range(4) for j in range(i+1, 4))

b = list(map(int, input("Enter state (e.g. 2,4,3,1): ").split(",")))
b = [x-1 for x in b]
step = 1

while cost(b):
    print("Step", step, ":", [x+1 for x in b], "Cost:", cost(b))
    best = b[:]
    for i in range(4):
        for r in range(4):
            t = b[:]
            t[i] = r
            if cost(t) < cost(best):
                best = t
    if best == b:
        b = [random.randrange(4) for _ in range(4)]
    else:
        b = best
    step += 1

print("Step", step, ":", [x+1 for x in b], "Cost:", cost(b))
print("Goal Reached!")
```
