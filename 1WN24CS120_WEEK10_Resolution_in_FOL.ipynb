```python

# Knowledge Base (CNF clauses)

KB = [
    ["~Eats(Anil,Peanuts)", "Food(Peanuts)"],
    ["~Alive(Anil)", "NotKilled(Anil)"],
    ["~Eats(Anil,Peanuts)", "~NotKilled(Anil)", "Food(Peanuts)"],
    ["~Food(Peanuts)", "Likes(John,Peanuts)"],
    ["Alive(Anil)"],
    ["Eats(Anil,Peanuts)"],
    ["~NotKilled(Anil)", "Alive(Anil)"]
]

# Negation of goal
KB.append(["~Likes(John,Peanuts)"])

# Resolution function
def resolve(c1, c2):
    for l1 in c1:
        for l2 in c2:

            if l1 == "~" + l2:
                new_clause = [
                    x for x in c1 if x != l1
                ] + [
                    x for x in c2 if x != l2
                ]

                return new_clause

            if l2 == "~" + l1:
                new_clause = [
                    x for x in c1 if x != l1
                ] + [
                    x for x in c2 if x != l2
                ]

                return new_clause

    return None


# Resolution
print("Hrishikesh shet - 1WN24CS120")
print("Knowledge Base:")
for clause in KB:
    print(clause)

print("\nResolving clauses...")

# Resolve Eats and Food
c1 = ["~Eats(Anil,Peanuts)", "Food(Peanuts)"]
c2 = ["Eats(Anil,Peanuts)"]

r1 = resolve(c1, c2)
print("Resolvent 1:", r1)

# Resolve Alive and NotKilled
c3 = ["~Alive(Anil)", "NotKilled(Anil)"]
c4 = ["Alive(Anil)"]

r2 = resolve(c3, c4)
print("Resolvent 2:", r2)

# Resolve Eats and NotKilled to derive Food
c5 = ["~Eats(Anil,Peanuts)", "~NotKilled(Anil)",
       "Food(Peanuts)"]

r3 = resolve(c5, c2)
print("Resolvent 3:", r3)

r4 = resolve(r3, r2)
print("Resolvent 4:", r4)

# Resolve Food with John Likes Food
c6 = ["~Food(Peanuts)", "Likes(John,Peanuts)"]

r5 = resolve(c6, r4)
print("Resolvent 5:", r5)

# Resolve with negated goal
goal = ["~Likes(John,Peanuts)"]

r6 = resolve(r5, goal)
print("Final resolvent:", r6)

if r6 == []:
    print("\nGoal proved: John likes peanuts.")
```
